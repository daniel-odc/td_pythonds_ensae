### Types de base

````python
# # Les types de base: int, float, str
a = 2
b = 3.14
c = 'texte'

print(f'a={a}, b={b}, c={c}')

# # variables non allouées à un type donné:
a = 'texte'  # a passe de int à str sans problème
print(f'new a={a}')
````

````python
# # Listes, tuples, dictionnaires et sets:

li = [1, 2, 3]  # `list`: []
print(f'type li: {type(li)}\n{li}')

tu = (1, 2, 3)  # `tuple`: ()
print(f'type lu: {type(tu)}\n{tu}')

di = {'a': 1, 'b': 2, 'c': 3}  # `dict`: {}
print(f'type di: {type(di)}\n{di}')

st = set([1, 2, 3])  # `set`: set([...]) ou {1, 2, 3}
print(f'type st: {type(st)}\n{st}')
````

````python
# # Listes:

li = [2, 4, 3]      # liste de 3 éléments
print(li)
print(f'1ère valeur: {li[0]}')        # l'index commence à 0

li.sort()           # tri la liste
print(li)

li.append(5)        # ajoute un élément
print(li)

li.extend([6, 7])   # ajoute une liste à la liste
print(li)

li.pop(4)           # retire le 4ème élément de la liste (et non l'élément 4 de la liste!)
print(li)

# # boucle sur les éléments:
for el in li:
    print(el)

# # boucle avec conservation des index:
for ix, el in enumerate(li):
    print(f'index {ix}, valeur: {el}')

# # Comprehension lists:
li = [i for i in range(10) if i%2 == 0]

````

````python
# # Dictionnaires:
# # my_dict = {key1: val1, key2: val2}

di = {'nom': 'Crusoé', 'prénom': 'Robinson', 'âge': 42}
print(di)

di['nom'] = 'Stévenin'  # les clés sont uniques
print(di)

prenom = di.get('prénom', '???')        # clé existante
adresse = di.get('adresse', '???')      # clé manquante
print(f'prénom:\t{prenom},\nadresse:\t{adresse}')
````

````python
# # Tuples:

tu = (49.4545, 2.4545)  # tuple à 2 éléments
print(f'{tu[0]}, {tu[1]}')

# tu[2] = 0               # le nombre d'éléments d'un tuple est invariable (pas de méthodes `append` ou `pop`)

# tu[1] = 0               # le contenu d'un tuple n'est pas modifiable
````

````python
# # Sets: Ensemble d'éléments uniques

st = {1, 3, 1, 2}
print(st)

st.add(4)
print(st)

st.pop()  # retire le premier élément
print(st)

st.remove(3)  # retire un élément donné
print(st)
````

## Mutabilité

En `Python`, seuls les types de base `int`, `float` et `str` ne sont pas mutables.

```python
a = 2
b = a
b = 3
print(f'a: {a}, b: {b}')  # le contenu de a n'a pas changé: une copie de a a été réalisée pour être alouée à b (idem si a est un float ou str)

li1 = [1, 2]
print(f'li1: {li1}')
li2 = li1
li2[0] = 0
print(f'li2: {li2}\nli1: {li1}')  # le contenu de li1 a été modifié lorsque li2 a été modifié

# # Solution: créer une copie de l'objet mutable:
li3 = li1.copy()
li3[0] = 30
print(f'li3: {li3}\nli1: {li1}')  # le contenu de li1 n'a pas été modifié lorsque li3 a été modifié
```

💡Les objets `Pandas` et `GeoPandas` qu'on verra par la suite sont mutables comme tous les objets `Python`

#### Fusions (`merge`)

![star schema](URL)

![img merge left etc.](URL)

````python
# # Exemple `merge`:
import pandas as pd

df_sale = pd.DataFrame({'year': [2012, 2014, 2013, 2014], 'month': [1, 4, 7, 10], 'sale': [55, 40, 84, 31]})

df_purchase = pd.DataFrame({'year': [2012, 2014, 2013, 2014], 'month': [1, 4, 7, 12], 'purchase': [45, 30, 90, 25]})

df_balance = df_sale.merge(df_purchase, on=['year', 'month'], how='left')  # jointure sur un ou plusieurs champs

print('Left join:')
display(df_balance)

df_balance = df_sale.merge(df_purchase, on=['year', 'month'], how='right')

print('Right join:')
display(df_balance)

df_balance = df_sale.merge(df_purchase, on=['year', 'month'], how='inner')

print('Inner join:')
display(df_balance)

df_balance = df_sale.merge(df_purchase, on=['year', 'month'], how='outer')

print('Outer join:')
display(df_balance)

````

````python
import geopandas as gpd
gdf_iris = gpd.read_parquet(r'c:\Users\dmarin\Dropbox\Pro\OpenDC\99 Projects\0001-Misc\Geopandas_cookbook\data\IRIS_France_odc.parquet')
gdf_iris.head()
````

````python
# # Points d'arrêt réseau TC Grand Lyon:
gdf_stops = gpd.read_file('https://data.grandlyon.com/geoserver/sytral/ows?SERVICE=WFS&VERSION=2.0.0&request=GetFeature&typename=sytral:tcl_sytral.tclarret&outputFormat=application/json&SRSNAME=EPSG:4171&sortBy=gid')
print(f"{len(gdf_stops.index):,} points d'arrêt")
print(f'Système de projection: {gdf_stops.crs}')
print(f'Dessertes uniques: {len(gdf_stops['desserte'].unique())}: {gdf_stops['desserte'].unique()}')
gdf_stops.head(2)
````

````python
# # Lignes de métro:
gdf_lines = gpd.read_file('https://data.grandlyon.com/geoserver/sytral/ows?SERVICE=WFS&VERSION=2.0.0&request=GetFeature&typename=sytral:tcl_sytral.tcllignemf_2_0_0&outputFormat=application/json&SRSNAME=EPSG:4171&sortBy=gid')
print(f"{len(gdf_lines.index):,} lignes de métro")
print(f'Système de projection: {gdf_lines.crs}')
gdf_lines.head(2)
````

````python
# # Entrées sorties du réseau de métro:
gdf_in_out = gpd.read_file('https://data.grandlyon.com/geoserver/sytral/ows?SERVICE=WFS&VERSION=2.0.0&request=GetFeature&typename=sytral:tcl_sytral.tclstation&outputFormat=application/json&SRSNAME=EPSG:4171&sortBy=gid')
print(f"{len(gdf_in_out.index):,} entrées / sorties de métro")
print(f'Système de projection: {gdf_in_out.crs}')
gdf_in_out.head(2)
````

````python
gdf_metro_stations = gdf_stops.merge(gdf_in_out, left_on='id', right_on='id_station', how='inner')
gdf_metro_stations.drop_duplicates(subset='id_x', inplace=True)
print(f'{len(gdf_metro_stations.index):,} stations de métro')
gdf_metro_stations.head()
````
